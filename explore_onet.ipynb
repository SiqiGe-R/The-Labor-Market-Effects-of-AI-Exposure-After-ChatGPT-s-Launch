# O*NET Task Statements — Exploratory Analysis

```python
import pandas as pd

df = pd.read_excel('Task_Statements.xlsx')
print('Columns:', df.columns.tolist())
print('Total rows:', len(df))
df.head()
```

```python
# Count unique SOC codes
n_soc = df['O*NET-SOC Code'].nunique()
print(f'Number of unique SOC codes: {n_soc}')
```

```python
# Distribution of task counts per occupation
task_counts = df.groupby('O*NET-SOC Code').size().rename('task_count')

print('Tasks per occupation — summary statistics:')
print(task_counts.describe().round(1))
print()
print(f'Most tasks:  {task_counts.idxmax()} ({task_counts.max()})')
print(f'Fewest tasks: {task_counts.idxmin()} ({task_counts.min()})')
```

```python
# List all occupations (SOC code + title)
soc_list = df[['O*NET-SOC Code', 'Title']].drop_duplicates(subset='O*NET-SOC Code').reset_index(drop=True)
print(f'First 20 occupations:')
soc_list.head(20)
```

```python
# Concatenate all task descriptions per occupation into a single string
# This is the 'merge task descriptions' step required before AI exposure scoring
merged = (
    df.groupby('O*NET-SOC Code')
    .agg(
        title=('Title', 'first'),
        task_count=('Task', 'count'),
        combined_tasks=('Task', lambda x: ' '.join(x.dropna().astype(str)))
    )
    .reset_index()
)

print(f'Occupations after merging: {len(merged)}')
merged.head()
```

```python
# Save merged output for downstream use
merged.to_csv('onet_merged_tasks.csv', index=False)
print('Saved to onet_merged_tasks.csv')
```
